```python
import pygame
from pygame.locals import *
import sys
import time
```

```python
#just initialising some variable we'll be using in our program
XO = 'x'
win = None
draw = False
w = 400 #width
h = 400 #height
bg_color = (255, 255, 255)
lines = (0,0,0)
board = [[None]*3,[None]*3,[None]*3]
```

```python
#configuring pygame necessities
pygame.init()
fps = 30
CLOCK = pygame.time.Clock()
screen = pygame.display.set_mode((w, h+100),0,32)
pygame.display.set_caption("Tic Tac Toe Game for AI Task 1")
```

```python
#loading the images
initiating_window = pygame.image.load("modified_cover-100x100.png")
X = pygame.image.load("Black-X-Letter-PNG-Clipart.png")
O = pygame.image.load("O-Letter-PNG-Pic.png")

initiating_window = pygame.transform.scale(initiating_window, (w, h + 100))
X = pygame.transform.scale(X, (80, 80))
O = pygame.transform.scale(O, (80, 80))
```

```python
#confihuring the basic canvas/ game screen on which the game will be played
def game_starting():
    screen.blit(initiating_window, (0, 0))
    pygame.display.update()
    time.sleep(1)
    screen.fill(bg_color)

    # Drawing vertical lines
    pygame.draw.line(screen,lines,(w/3,0),(w/3, h),7)
    pygame.draw.line(screen,lines,(w/3*2,0),(w/3*2, h),7)
    # Drawing horizontal lines
    pygame.draw.line(screen,lines,(0,h/3),(w, h/3),7)
    pygame.draw.line(screen,lines,(0,h/3*2),(w, h/3*2),7)
    draw_status()
```

```python
def draw_status():
    global draw

    if win is None:
        message = XO.upper() + "'s Turn To Play!"
    else:
        message = win.upper() + " Won!"
    if draw:
        message = 'Game Draw!'

    font = pygame.font.Font(None, 30)
    text = font.render(message, 1, (255, 255, 255))

    
    screen.fill ((0, 0, 0), (0, 400, 500, 100))
    text_rect = text.get_rect(center=(w/2, 500-50))
    screen.blit(text, text_rect)
    pygame.display.update()
```

```python
def check_win():
    global board, win,draw

    # check if three in a row
    for row in range (0,3):
        if ((board [row][0] == board[row][1] == board[row][2]) and(board [row][0] is not None)):
            win = board[row][0]
            pygame.draw.line(screen, (250,0,0), (0, (row + 1)*h/3 -h/6),\
                              (w, (row + 1)*h/3 - h/6 ), 4)
            break

    # check if three in a column
    for col in range (0, 3):
        if (board[0][col] == board[1][col] == board[2][col]) and (board[0][col] is not None):
            win = board[0][col]
            pygame.draw.line (screen, (250,0,0),((col + 1)* w/3 - w/6, 0),\
                          ((col + 1)* w/3 - w/6, h), 4)
            break

    # check for diagonal winners
    #case 1
    if (board[0][0] == board[1][1] == board[2][2]) and (board[0][0] is not None):
        win = board[0][0]
        pygame.draw.line (screen, (250,70,70), (50, 50), (350, 350), 4)
    #case 2
    if (board[0][2] == board[1][1] == board[2][0]) and (board[0][2] is not None):
        win = board[0][2]
        pygame.draw.line (screen, (250,70,70), (350, 50), (50, 350), 4)

    if(all([all(row) for row in board]) and win is None ):
        draw = True
    draw_status()
```

```python
def drawXO(row,col):
    global board,XO
    

    if col==1:
        posy = 30
    if col==2:
        posy = h/3 + 30
    if col==3:
        posy = h/3*2 + 30
    if row==1:
        posx = 30
    if row==2:
        posx = w/3 + 30
    if row==3:
        posx = w/3*2 + 30
    board[row-1][col-1] = XO
    if(XO == 'x'):
        screen.blit(X,(posy,posx))
        XO= 'o'
    else:
        screen.blit(O,(posy,posx))
        XO= 'x'
    pygame.display.update()
```

```python
def userClick():
    x,y = pygame.mouse.get_pos()
    if(x<w/3):
        col = 1
    elif (x<w/3*2):
        col = 2
    elif(x<w):
        col = 3
    else:
        col = None

    if(y<h/3):
        row = 1
    elif (y<h/3*2):
        row = 2
    elif(y<h):
        row = 3
    else:
        row = None

    if(row and col and board[row-1][col-1] is None):
        global XO

        drawXO(row,col)
        check_win()
```

```python
def reset_game():
    global board, win,XO, draw
    time.sleep(3)
    XO = 'x'
    draw = False
    game_starting()
    win=None
    board = [[None]*3,[None]*3,[None]*3]
```

```python
game_starting()
while(True):
    for event in pygame.event.get():
        if event.type == QUIT:
            pygame.quit()
            sys.exit()
        elif event.type == MOUSEBUTTONDOWN:
            userClick()
            if(win or draw):
                reset_game()

    pygame.display.update()
    CLOCK.tick(fps)
```
