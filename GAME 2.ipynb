```python
import pygame
from pygame.locals import *
import sys
import time
import random

pygame.init()

w = 800
h = 600
window = pygame.display.set_mode((w,h))
pygame.display.set_caption('Snake Game for AI Task 1')
font=pygame.font.SysFont('comicsans',40)
```

```python
def main_menu():
    while 1:
        for event in pygame.event.get():
            if event.type == pygame.QUIT:
                pygame.quit()
                sys.exit()
            if event.type == pygame.MOUSEBUTTONDOWN:
                main()
        window.fill((0,0,0))
        main_menu_message = font.render('Click anywhere to start' , True , (255,255,255))
        font_pos = main_menu_message.get_rect(center=(w//2, h//2))
        window.blit(main_menu_message , font_pos)
        pygame.display.update()
```

```python
def game_over(score):
        while 1:
            for event in pygame.event.get():
                if event.type == pygame.QUIT:
                    pygame.quit()
                    sys.exit()
            window.fill((0,0,0))
            game_over_message = font.render('You  have lost :(' , True , (255,0,0))
            game_over_score = font.render(f'Your score was {score}' , True , (255,255,255))

            font_pos_message = game_over_message.get_rect(center=(w//2, h//2))
            font_pos_score = game_over_score.get_rect(center=(w//2, h//2+40))
            window.blit(game_over_message , font_pos_message)
            window.blit(game_over_score , font_pos_score)
            pygame.display.update()
            time.sleep(3)
            main_menu()
```

```python
def main():
    CLOCK = pygame.time.Clock()
    direction = 'right'
    snake_pos=[200,70]
    snake_body=[[200,70] , [200-10 , 70] , [200-(2*10),70]]
    fruit_pos = [0,0]
    fruit_spawn = True
    score=0
    CLOCK = pygame.time.Clock()
    #game loop
    while 1:
        for event in pygame.event.get():
            if event.type == pygame.QUIT:
                pygame.quit()
                sys.exit()

            keys= pygame.key.get_pressed()
            if (keys[pygame.K_w] or keys[pygame.K_UP]) and direction != 'down':
                direction = 'up'
            if (keys[pygame.K_s] or keys[pygame.K_DOWN]) and direction != 'up':
                direction = 'down'
            if (keys[pygame.K_d] or keys[pygame.K_RIGHT]) and direction != 'left':
                direction = 'right'
            if (keys[pygame.K_a] or keys[pygame.K_LEFT]) and direction != 'right':
                direction = 'left'

        window.fill((0,0,0))
        for square in snake_body:
            pygame.draw.rect(window ,(255, 255, 0), (square[0],square[1],10,10))
        if direction == 'right':
            snake_pos[0] += 10
        elif direction == 'left':
            snake_pos[0] -= 10
        elif direction == 'up':
            snake_pos[1] -= 10
        elif direction == 'down':
            snake_pos[1] += 10

        snake_body.append(list(snake_pos))

        if fruit_spawn:
            fruit_pos = [random.randrange(40,w-40),random.randrange(40,h-40)]
            fruit_spawn = False

        pygame.draw.rect(window ,(138,43,226),(fruit_pos[0],fruit_pos[1],10,10))

        score_font = font.render(f'{score}' , True , (255,255,255))
        font_pos = score_font.get_rect(center=(w//2-40 , 30))
        window.blit(score_font , font_pos)

        if pygame.Rect(snake_pos[0],snake_pos[1],10,10).colliderect(pygame.Rect(fruit_pos[0],fruit_pos[1],10,10)):
            fruit_spawn=True
            score += 10
        else:
            snake_body.pop(0)

        for square in snake_body[:-1]:
            if pygame.Rect(square[0],square[1],10,10).colliderect(pygame.Rect(snake_pos[0],snake_pos[1],10,10)):
                game_over(score)

        if snake_pos[0]+10 <=0 or snake_pos[0] >= w:
            game_over(score)
        if snake_pos[1]+10 <=0 or snake_pos[1] >= h:
            game_over(score)
        pygame.display.update()

        CLOCK.tick(25)

#caliing the main function
main_menu()
```
